## Q1. Check if Lionel Messi has more than 10 achievements

```python
players = [
    ["Serena Williams", "Tennis", 23],
    ["Lionel Messi", "Soccer", 7],
    ["Michael Phelps", "Swimming", 23],
    ["Usain Bolt", "Athletics", 8],
    ["Roger Federer", "Tennis", 20],
    ["Cristiano Ronaldo", "Soccer", 5]
]

for player in players:
    if player[0] == "Lionel Messi":
        if player[2] > 10:
            print("Player Name:", player[0])
            print("Sport:", player[1])
            print("Achievements:", player[2])
        else:
            print("Lionel Messi does not have more than 10 achievements.")
```

## Q4. Check even/odd, palindrome, and divisibility by the sum of digits

```python
num = int(input("Enter a number between 1 and 9999: "))

# Check even or odd
if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

# Check palindrome
if str(num) == str(num)[::-1]:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")

# Find sum of digits
temp = num
sum_digits = 0

while temp > 0:
    digit = temp % 10
    sum_digits = sum_digits + digit
    temp = temp // 10

print("Sum of digits:", sum_digits)

# Check divisibility by sum of digits
if sum_digits != 0 and num % sum_digits == 0:
    print("The number is divisible by the sum of its digits.")
else:
    print("The number is not divisible by the sum of its digits.")
```

## Q7. Copy strings to new_squares and stop if the value is not 'orange'

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']
new_squares = []

for value in squares:
    if value == 'orange':
        new_squares.append(value)
    else:
        break

print(new_squares)
```

## Q8. Use a while loop to find animal names made of 7 letters

```python
Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]
new_animals = []
i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        new_animals.append(Animals[i])
    i = i + 1

print(new_animals)
```
