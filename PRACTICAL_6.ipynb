## Problem 1 – Fix buggy code using super()

```python
class Person:
    # Constructor
    def __init__(self, name, id):
        self.name = name
        self.id = id

    # To display person details
    def Display(self):
        print(self.name, self.id)


class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


Emp_details = Emp("Mayank", 103)

# Calling parent class function
Emp_details.Display()

# Calling child class function
Emp_details.Print()
```

## Problem 2 – Anagram Check

```python
def check_anagram(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    if sorted(str1) == sorted(str2):
        print("The strings are Anagrams.")
    else:
        print("The strings are not Anagrams.")


str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

check_anagram(str1, str2)
```

## Problem 3 – Armstrong Number

```python
num = int(input("Enter a number: "))

original_num = num
digits = len(str(num))
digit_sum = 0

while num > 0:
    digit = num % 10
    digit_sum = digit_sum + digit ** digits
    num = num // 10

if digit_sum == original_num:
    print("The number is an Armstrong number.")
else:
    print("The number is not an Armstrong number.")
```

## Problem 4 – Classes and Objects
Scenario: Car Dealership's Inventory Management System

```python
class Vehicle:
    # Constructor
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "white"
        self.seating_capacity = 0

    # Method to assign seating capacity
    def assign_seating_capacity(self, seating_capacity):
        self.seating_capacity = seating_capacity

    # Method to display all properties
    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)


# Creating first vehicle object
car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

# Creating second vehicle object
car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)

# Displaying properties
print("Vehicle 1:")
car1.display_properties()

print("\nVehicle 2:")
car2.display_properties()
```
