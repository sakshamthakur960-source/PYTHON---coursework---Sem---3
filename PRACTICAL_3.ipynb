## Dictionaries and Tuples

## Part A – Dictionaries

```python
# 1. Create an empty dictionary
inventory = {}

# 2. Store the first product details in variables
product1_name = "Mobile Phone"
product1_quantity = 5
product1_price = 20000
product1_releaseYear = 2020

# 3. Add details in inventory
inventory["ProductNo1"] = {
    "productName": product1_name,
    "quantity": product1_quantity,
    "price": product1_price,
    "releaseYear": product1_releaseYear
}

# 4. Store the second product details in variables
product2_name = "Laptop"
product2_quantity = 3
product2_price = 50000
product2_releaseYear = 2022

# 5. Add the second product to inventory
inventory["ProductNo2"] = {
    "productName": product2_name,
    "quantity": product2_quantity,
    "price": product2_price,
    "releaseYear": product2_releaseYear
}

# 6. Display the products present in inventory
print("Products in Inventory:")
for product, details in inventory.items():
    print(product, ":", details)

# 7. Check if release year is present
print("\nChecking release year:")
if "releaseYear" in inventory["ProductNo1"]:
    print("ProductNo1 release year is present.")
if "releaseYear" in inventory["ProductNo2"]:
    print("ProductNo2 release year is present.")

# 8. Delete release year of both products
del inventory["ProductNo1"]["releaseYear"]
del inventory["ProductNo2"]["releaseYear"]

print("\nInventory after deleting release years:")
for product, details in inventory.items():
    print(product, ":", details)
```

## Part B – Tuples

```python
# Create a tuple of product prices
prices = (250, 300, 150, 400, 100, 350, 200)

# Find the highest and lowest price
highest_price = max(prices)
lowest_price = min(prices)

print("Highest Price:", highest_price)
print("Lowest Price:", lowest_price)

# Calculate the total sum
total = sum(prices)
print("Total Sum:", total)

# Convert tuple into a sorted list
sorted_prices = sorted(prices)
print("Sorted Prices:", sorted_prices)

# Try to modify an element of the tuple
try:
    prices[0] = 500
except TypeError as e:
    print("Error:", e)
```
