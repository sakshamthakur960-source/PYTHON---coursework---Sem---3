```python
# Q1 Create an empty list
shopping_list = []
print(shopping_list)

# Q2 Store the items in the shopping list
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]
print(shopping_list)

# Q3 Add Football to the shopping list
shopping_list.append("Football")
print(shopping_list)

# Q4 Print the last item
print("Last item:", shopping_list[-1])

# Q5 Print the entire shopping list
print("Shopping List:", shopping_list)

# Q6 Print important items: Laptop and Shoes
print("Important items:", shopping_list[1], "and", shopping_list[2])

# Q7 Change Pen to Notebook
shopping_list[3] = "Notebook"
print("Updated Shopping List:", shopping_list)

# Q8 Delete the least important item and print the list
# Assuming Clothes is the least important item:
shopping_list.remove("Clothes")
print("Final Shopping List:", shopping_list)
```
